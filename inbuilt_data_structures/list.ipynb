```python
pair2.append([3, 'd'])        # add an inner list at the end

pair2.insert(1, [0, 'z'])     # insert at index 1

pair2.pop(2)                  # remove and return item at index 2

del pair2[2]                  # remove item at index 2

pair2.remove([1, 'b'])        # remove by matching value

pair2[0][1] = 'x'             # changes [1, 'a'] to [1, 'x']
```

```python
n = int(input())
lst = []

for _ in range(n):
    command = input().split()

    if command[0] == "insert":
        lst.insert(int(command[1]), int(command[2]))

    elif command[0] == "print":
        print(lst)

    elif command[0] == "remove":
        lst.remove(int(command[1]))

    elif command[0] == "append":
        lst.append(int(command[1]))

    elif command[0] == "sort":
        lst.sort()

    elif command[0] == "pop":
        lst.pop()

    elif command[0] == "reverse":
        lst.reverse()
```
