```python
# set membership test

my_set = {1,2,3,4,5}
print(3 in my_set)
print("n" in my_set)
```

```python
set1 = {1,2,3,4,5,6,7}
set2 = {5,6,7,8,9,10}

# union
union_set = set1.union(set2)
print(union_set)

# intersection
intersection_set = set1.intersection(set2)
print(intersection_set)


set1.intersection_update(set2)
print(set1)
```

```python
set1 = {1,2,3,4,5,6,7}
set2 = {5,6,7,8,9,10}


difference_set = set1.difference(set2)
print(difference_set)

set1.difference_update(set2)
print(set1)
```

```python
#symmetric difference

set1 = {1,2,3,4,5,6,7}
set2 = {5,6,7,8,9,10}

symmetric_diff_set = set1.symmetric_difference(set2)
print(symmetric_diff_set)

set3 = set1.symmetric_difference_update(set2)
print(set1)
print(set3)
```

```python
# sets method 

set1 = {1,2,3}
set2 = {1,2,3,4,5,6}

set3 = {"a","b","c","d"}
set4 = {"b","d"}

# is subset 
print(set1.issubset(set2))

# is superset
print(set3.issuperset(set4))
```

```python
lst = [1,2,3,4,4,5,6]
set1 = set(lst)
print(set1)
```

```python
## counting unique words in  text 
text = "My name is Ajay Hegde, and I am learning coding"
words = text.split()
print(words)

set1 = set(text)
print(set1)

set2 = set(words)
print(set2)

print(sorted(set1))
print(sorted(set2))
```
