```python
empty_tuple = ()
print(empty_tuple)
print(type(empty_tuple))
```

```python
lst = list()
print(type(lst))

tpl = tuple()
print(type(tpl))
```

```python
number = (1,2,3,4,5,6)
print(type(number))

num = tuple([1,2,3,4,5])
num 
```

```python
## Accessing tuple elements
number[0]
number[0:4]
```

```python
number = (1,2,3,4,5)
mixed_tuple =(1,2,"Hello world", 3.17)
print(number, mixed_tuple)
```

```python
#concatenation
n = number + mixed_tuple
print(n)
```

```python
mul = mixed_tuple * 3
print(mul)
```

```python
lst = [1,2,3,4]
lst[1] = "Ajay"
print(lst)

#immutble
tpl = (1,2,3)
tpl[1] = "Ajay"
print(tpl)
```

```python
## tuples method
print(number.count(1))
print(number.index(4)) # index of the value passed as argument
```

```python
## packing and unpacking 
#packing
packed_tuple = 1, "Hello", 3.18, 6
print(packed_tuple)

#unpacking
a,b,c,d = packed_tuple
print(a)
print(b)
print(c)
print(d)
```

```python
## unpacking with *

number = (1,2,3,4,5,6,6,7,"Ajay","suii")
first , *middle, last = number
print(first)
print(middle)
print(*middle)
print(last)
```

```python
## Nested Tuple

#nested list 
lst = [[1,2,3], ["Ajay", "Bhoonki", "Shweta"], (3.14, "suii", "Rax", 5.5)]
print(lst)
print(lst[2][2])
```

```python
nested_tuple = ((1,2,3), ("Ajay","Rax", 3.19))
print(nested_tuple)

print(nested_tuple[1][1])
```

```python
for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item, end=" ")
```
