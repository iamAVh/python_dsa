```python
text = "hello world"

print(text)
print(len(text))                 # length
print(text[0])                   # indexing
print(text[-1])                  # last character
print(text[0:5])                 # slicing
print(text.upper())              # uppercase
print(text.lower())              # lowercase
print(text.capitalize())         # first letter capital
print(text.title())              # first letter of each word capital
print(text.swapcase())           # upper to lower and lower to upper

print(text.strip())              # remove spaces from both sides
print(text.lstrip())             # remove left spaces
print(text.rstrip())             # remove right spaces

print(text.replace("world", "python"))   # replace word
print(text.split())              # split into list
print("-".join(text.split()))    # join list into string

print(text.find("world"))        # index of substring
print(text.index("world"))       # same as find, but gives error if not found
print(text.count("l"))           # count characters
print(text.startswith("hello"))  # check start
print(text.endswith("world"))    # check end

print(text.isalpha())            # only letters?
print(text.isdigit())            # only digits?
print(text.isalnum())            # letters + digits?
print(text.isspace())            # only spaces?
print(text.islower())            # all lowercase?
print(text.isupper())            # all uppercase?

print("hello" in text)           # membership check
print(text + " !!!")             # concatenation
print(text * 3)                  # repetition
```
